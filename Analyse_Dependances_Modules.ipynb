```python
import pandas as pd
import os

# Define unique_modules_precedent before using it in the loop
unique_modules_precedent = df_dependance_sequence_normalized['module_precedent'].unique()

print("Generating session sequences for all unique modules...")

# Initialize a set to store indices of used dependencies from the original DataFrame
all_used_dependency_original_indices = set()

def generate_module_sequence(module_name, df_dependance_sequence_normalized, all_used_dependency_original_indices_ref):
    # Filter dependencies for the current module
    df_module_dependencies = df_dependance_sequence_normalized[
        (df_dependance_sequence_normalized['module_precedent'] == module_name) |
        (df_dependance_sequence_normalized['module_suivant'] == module_name)
    ].copy() # Use .copy() to avoid SettingWithCopyWarning

    total_rows_for_module = len(df_module_dependencies)

    if df_module_dependencies.empty:
        return pd.DataFrame(columns=['type', 'numero']), total_rows_for_module, 0

    # Convert 'numero_precedent' and 'numero_suivant' to numeric
    df_module_dependencies['numero_precedent'] = pd.to_numeric(df_module_dependencies['numero_precedent'], errors='coerce')
    df_module_dependencies['numero_suivant'] = pd.to_numeric(df_module_dependencies['numero_suivant'], errors='coerce')

    # Drop rows where conversion resulted in NaN
    df_module_dependencies.dropna(subset=['numero_precedent', 'numero_suivant'], inplace=True)

    if df_module_dependencies.empty:
        return pd.DataFrame(columns=['type', 'numero']), total_rows_for_module, 0

    # Sort dependencies to ensure consistent order
    df_module_dependencies.sort_values(
        by=['type_precedent', 'numero_precedent', 'type_suivant', 'numero_suivant'],
        inplace=True
    )

    sequence = []
    # Try to find a starting point 'CM 1'
    start_session = df_module_dependencies[
        (df_module_dependencies['type_precedent'] == 'CM') &
        (df_module_dependencies['numero_precedent'] == 1)
    ]

    current_session = None
    if not start_session.empty:
        # If CM 1 is a precedent, it's a valid starting point for the sequence
        current_session = {'type': 'CM', 'numero': 1}
        sequence.append(current_session)
    elif not df_module_dependencies[
        (df_module_dependencies['type_suivant'] == 'CM') &
        (df_module_dependencies['numero_suivant'] == 1)
    ].empty:
        # If CM 1 is a 'suivant' in any dependency, it implies it exists but might not be a 'precedent' in the current filtered df
        # This could happen if CM 1 is the absolute beginning for the module and has no precedent, so we just assume it as a start
        # However, for the purpose of dependency chaining, we need it to be a 'precedent' for the loop to start.
        # If CM 1 is not a precedent, the chain won't start from it directly using the current logic.
        # Let's check for any 'CM 1' in the entire module's dependencies that can act as a starting point.

        # More robust start: Find all unique sessions (type, numero) for the module
        all_module_sessions = pd.concat([
            df_module_dependencies[['type_precedent', 'numero_precedent']].rename(columns={'type_precedent':'type', 'numero_precedent':'numero'}),
            df_module_dependencies[['type_suivant', 'numero_suivant']].rename(columns={'type_suivant':'type', 'numero_suivant':'numero'})
        ]).drop_duplicates().sort_values(by=['type', 'numero']).to_dict('records')

        # Find the earliest 'CM 1' if it exists as any session type
        for session in all_module_sessions:
            if session['type'] == 'CM' and session['numero'] == 1:
                current_session = session
                sequence.append(current_session)
                break

    if current_session is None:
        # If no CM 1 is found or cannot be established as a starting point, return an empty sequence or log this.
        return pd.DataFrame(columns=['type', 'numero']), total_rows_for_module, 0


    # This set will keep track of visited sessions to detect cycles and avoid infinite loops
    visited_sessions = set()
    visited_sessions.add(f"{current_session['type']}_{current_session['numero']}")

    # Re-initialize current_session for the loop from the actual start
    sequence = [current_session]
    rows_used_in_sequence = 0 # This will count the number of dependency links used

    while True:
        # Find the next dependency in the chain where the current session is the precedent
        next_dependency_options = df_module_dependencies[
            (df_module_dependencies['type_precedent'] == current_session['type']) &
            (df_module_dependencies['numero_precedent'] == current_session['numero'])
        ].sort_values(by=['type_suivant', 'numero_suivant'])

        if next_dependency_options.empty:
            # No more dependencies found, sequence ends
            break

        # Prioritize 'CM' and then 'TD'/'TP' for the next step if multiple options exist.
        # For simplicity, we just take the first sorted option if multiple exist.
        next_session_info = next_dependency_options.iloc[0]

        next_type = next_session_info['type_suivant']
        next_numero = next_session_info['numero_suivant']

        next_session_key = f"{next_type}_{next_numero}"
        if next_session_key in visited_sessions:
            # Cycle detected or already processed this session in the current path. Terminate.
            break

        current_session = {'type': next_type, 'numero': next_numero}
        sequence.append(current_session)
        visited_sessions.add(next_session_key)
        rows_used_in_sequence += 1 # Increment for each dependency link successfully traversed

        # Add the original index of the used dependency row to the global set
        # next_session_info.name holds the original index in df_dependance_sequence_normalized
        all_used_dependency_original_indices_ref.add(next_session_info.name)

    return pd.DataFrame(sequence), total_rows_for_module, rows_used_in_sequence


# Initialize overall counters and lists to capture module information
total_dependencies_used_across_all_sequences = 0
total_rows_in_original_df = len(df_dependance_sequence_normalized)
modules_without_cm1_start_sequence = []

# Define output directory for CSVs
output_dir = 'module_sequences_csv'
os.makedirs(output_dir, exist_ok=True)

# Iterate through each unique module and generate its session sequence
for module in unique_modules_precedent:
    module_sequence_df, total_module_deps_count, used_deps_in_sequence_count = generate_module_sequence(module, df_dependance_sequence_normalized, all_used_dependency_original_indices)

    print(f"Séquence des séances pour le module {module}:")
    print(f"  Total de lignes concernant ce module dans df_dependance_sequence_normalized: {total_module_deps_count}")

    is_cm_1_start = False
    if not module_sequence_df.empty:
        first_session = module_sequence_df.iloc[0]
        if first_session['type'] == 'CM' and first_session['numero'] == 1:
            is_cm_1_start = True

    if not module_sequence_df.empty and is_cm_1_start:
        print(module_sequence_df.to_string(index=False))
        print(f"  Nombre de lignes de dépendance utilisées dans la construction de la séquence: {used_deps_in_sequence_count}")
        total_dependencies_used_across_all_sequences += used_deps_in_sequence_count

        # Save the sequence to a CSV file
        filename = f"{output_dir}/{module}_sequence.csv"
        module_sequence_df.to_csv(filename, index=False)
        print(f"  Séquence sauvegardée dans '{filename}'")

    else:
        print("  Aucune séquence trouvée ou ne commence pas par CM 1. Seul le nombre total de lignes pour le module est affiché.")
        modules_without_cm1_start_sequence.append(module) # Capture modules that failed CM1 start

    print("----------------------------------------------------------------------\n")

# After the loop, print the overall comparison
print("--- Récapitulatif global ---")
print(f"Total de lignes dans df_dependance_sequence_normalized: {total_rows_in_original_df}")
print(f"Total de lignes de dépendance utilisées dans toutes les séquences construites: {total_dependencies_used_across_all_sequences}")
if total_dependencies_used_across_all_sequences == total_rows_in_original_df:
    print("Toutes les lignes de df_dependance_sequence_normalized ont été utilisées dans la construction des séquences.")
else:
    print(f"ATTENTION: {total_rows_in_original_df - total_dependencies_used_across_all_sequences} lignes de df_dependance_sequence_normalized n'ont PAS été utilisées dans la construction des séquences."
          " Cela peut être dû à des dépendances non liées, des modules sans point de départ 'CM 1', ou des cycles qui ont interrompu la construction de la séquence.")

    # Display the unused lines
    all_original_indices = set(df_dependance_sequence_normalized.index)
    unused_original_indices = all_original_indices - all_used_dependency_original_indices

    print(f"--- Lignes non utilisées dans df_dependance_sequence_normalized ({len(unused_original_indices)} lignes) ---")
    if not unused_original_indices:
        print("Aucune ligne non utilisée trouvée.")
    else:
        # Sort indices for consistent output
        sorted_unused_indices = sorted(list(unused_original_indices))
        df_unused_total = df_dependance_sequence_normalized.loc[sorted_unused_indices].copy()
        print(df_unused_total.to_string())

        # Now, specifically identify the "other 4 lines" - these are unused lines that belong to modules *not* in `modules_without_cm1_start_sequence`
        unused_module_names_in_total_unused = df_unused_total['module_precedent'].unique()

        df_unused_from_cm1_successful_modules_list = []
        for module_name in unused_module_names_in_total_unused:
            if module_name not in modules_without_cm1_start_sequence:
                # These are modules that HAD a CM1-starting sequence, but still have unused dependencies
                df_unused_from_cm1_successful_modules_list.append(df_unused_total[df_unused_total['module_precedent'] == module_name])

        if df_unused_from_cm1_successful_modules_list:
            df_unused_from_cm1_successful_modules = pd.concat(df_unused_from_cm1_successful_modules_list)
            print(f"\n--- Lignes non utilisées provenant de modules AVEC une séquence démarrant par CM 1 ({len(df_unused_from_cm1_successful_modules)} lignes) ---")
            print("Ces lignes n'ont pas été intégrées dans la séquence linéaire construite, potentiellement à cause de chemins alternatifs ou de sous-graphes déconnectés au sein du module.")
            print(df_unused_from_cm1_successful_modules.to_string())
        else:
            print("\n--- Aucune ligne non utilisée provenant de modules AVEC une séquence démarrant par CM 1 ---")
```

```python
import pandas as pd

def generate_specific_sequence(module_name, start_type, start_numero, df_dependance_sequence_normalized):
    """
    Construit une séquence de sessions pour un module donné à partir d'un point de départ spécifié.
    Si aucune dépendance explicite n'est trouvée, essaie d'incrémenter le numéro de session de 1.

    Args:
        module_name (str): Le nom du module (e.g., 'PROJ932_IDU').
        start_type (str): Le type de la session de départ (e.g., 'TP', 'TD', 'CM').
        start_numero (int): Le numéro de la session de départ (e.g., 1, 5).
        df_dependance_sequence_normalized (pd.DataFrame): Le DataFrame contenant toutes les dépendances.

    Returns:
        pd.DataFrame: Un DataFrame avec les colonnes 'type' et 'numero' représentant la séquence,
                      ou un DataFrame vide si aucune séquence ne peut être construite.
    """
    df_module_dependencies = df_dependance_sequence_normalized[
        (df_dependance_sequence_normalized['module_precedent'] == module_name) |
        (df_dependance_sequence_normalized['module_suivant'] == module_name)
    ].copy()

    if df_module_dependencies.empty:
        print(f"Aucune dépendance trouvée pour le module {module_name}.")
        return pd.DataFrame(columns=['type', 'numero'])

    # Convert 'numero_precedent' and 'numero_suivant' to numeric for comparison
    df_module_dependencies['numero_precedent'] = pd.to_numeric(df_module_dependencies['numero_precedent'], errors='coerce').astype('Int64')
    df_module_dependencies['numero_suivant'] = pd.to_numeric(df_module_dependencies['numero_suivant'], errors='coerce').astype('Int64')

    # Drop rows where conversion resulted in NaN
    df_module_dependencies.dropna(subset=['numero_precedent', 'numero_suivant'], inplace=True)

    if df_module_dependencies.empty:
        print(f"Aucune dépendance valide après conversion numérique pour le module {module_name}.")
        return pd.DataFrame(columns=['type', 'numero'])

    sequence = []
    current_session = {'type': start_type, 'numero': start_numero}

    # Pre-calculate all unique sessions for the module to check for sequential existence
    all_module_sessions_flat = pd.concat([
        df_module_dependencies[['type_precedent', 'numero_precedent']].rename(columns={'type_precedent':'type', 'numero_precedent':'numero'}),
        df_module_dependencies[['type_suivant', 'numero_suivant']].rename(columns={'type_suivant':'type', 'numero_suivant':'numero'})
    ]).drop_duplicates()

    # Verify if the starting session actually exists in the module's overall sessions
    is_valid_start = all_module_sessions_flat[
        (all_module_sessions_flat['type'] == start_type) &
        (all_module_sessions_flat['numero'] == start_numero)
    ]

    if is_valid_start.empty:
        print(f"Le point de départ spécifié ({start_type} {start_numero}) n'est pas trouvé dans les sessions du module {module_name}.")
        return pd.DataFrame(columns=['type', 'numero'])

    sequence.append(current_session)
    visited_sessions = {f"{current_session['type']}_{current_session['numero']}"}

    while True:
        next_session = None

        # 1. Try to find an explicit dependency
        next_dependency_options = df_module_dependencies[
            (df_module_dependencies['type_precedent'] == current_session['type']) &
            (df_module_dependencies['numero_precedent'] == current_session['numero'])
        ].sort_values(by=['type_suivant', 'numero_suivant'])

        if not next_dependency_options.empty:
            # Found an explicit dependency, use the first one
            next_session_info = next_dependency_options.iloc[0]
            next_type = next_session_info['type_suivant']
            next_numero = next_session_info['numero_suivant']
            next_session = {'type': next_type, 'numero': next_numero}
        else:
            # 2. If no explicit dependency, try to increment 'numero' by 1 for the same type
            potential_next_numero = current_session['numero'] + 1

            sequential_next_exists = all_module_sessions_flat[
                (all_module_sessions_flat['type'] == current_session['type']) &
                (all_module_sessions_flat['numero'] == potential_next_numero)
            ]

            if not sequential_next_exists.empty:
                # Sequential next session exists and can be considered
                next_session = {'type': current_session['type'], 'numero': potential_next_numero}
            else:
                # No explicit dependency and no sequential increment possible, so break
                break

        if next_session is None: # Should not happen if break conditions are well handled, but for safety.
            break

        next_session_key = f"{next_session['type']}_{next_session['numero']}"
        if next_session_key in visited_sessions:
            # Cycle detected or already processed this session. Terminate.
            break

        current_session = next_session
        sequence.append(current_session)
        visited_sessions.add(next_session_key)

    return pd.DataFrame(sequence)

# --- Génération des séquences demandées par l'utilisateur ---

# Séquence pour "PROJ932_IDU" à partir de "TP 1"
module_proj932 = "PROJ932_IDU"
start_proj932_type = "TP"
start_proj932_numero = 1
sequence_proj932 = generate_specific_sequence(module_proj932, start_proj932_type, start_proj932_numero, df_dependance_sequence_normalized)

print(f"Séquence pour le module {module_proj932} à partir de {start_proj932_type} {start_proj932_numero}:")
if not sequence_proj932.empty:
    print(sequence_proj932.to_string(index=False))
else:
    print("  Aucune séquence n'a pu être construite pour ce point de départ.")

print("\n----------------------------------------------------------------------\n")

# Séquence pour "INFO631_IDU" à partir de "TD 1"
module_info631 = "INFO631_IDU"
start_info631_type = "TD"
start_info631_numero = 1
sequence_info631 = generate_specific_sequence(module_info631, start_info631_type, start_info631_numero, df_dependance_sequence_normalized)

print(f"Séquence pour le module {module_info631} à partir de {start_info631_type} {start_info631_numero}:")
if not sequence_info631.empty:
    print(sequence_info631.to_string(index=False))
else:
    print("  Aucune séquence n'a pu être construite pour ce point de départ.")
```
